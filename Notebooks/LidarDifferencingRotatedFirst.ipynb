```python
from osgeo import gdal
import math

def rotate_raster(input_path, output_path, angle_degrees, pivot_x, pivot_y):
    # 1. Open the source dataset
    src_ds = gdal.Open(input_path)
    if not src_ds:
        print("Failed to open the input file.")
        return

    # Extract original geotransform
    gt = src_ds.GetGeoTransform()
    
    # Extract native resolution to enforce it in the final warped output
    # Absolute value is used for y_res because it is traditionally negative
    orig_x_res = abs(gt[1])
    orig_y_res = abs(gt[5])

    # 2. Compute the rotation matrix around the pivot point
    radians = math.radians(angle_degrees)
    cos_a = math.cos(radians)
    sin_a = math.sin(radians)

    # Matrix translation accounting for the raster's existing origin (gt[0], gt[3])
    xoff = pivot_x + (gt[0] - pivot_x) * cos_a - (gt[3] - pivot_y) * sin_a
    yoff = pivot_y + (gt[0] - pivot_x) * sin_a + (gt[3] - pivot_y) * cos_a

    # Apply standard 2D affine rotation formulas to the scale components
    gt_new = [
        xoff,
        gt[1] * cos_a,   # New X-scale (pixel width)
        gt[1] * sin_a,   # X-rotation / shearing coefficient
        yoff,
        -gt[5] * sin_a,  # Y-rotation / shearing coefficient
        gt[5] * cos_a    # New Y-scale (pixel height)
    ]
    
    print("Original GT:", gt)
    print("Rotated GT: ", gt_new)

    # 3. Apply the new geotransform to an in-memory virtual dataset
    tmp_ds = gdal.GetDriverByName('MEM').CreateCopy('', src_ds)
    tmp_ds.SetGeoTransform(gt_new)

    # 4. Correctly configure GDAL Warp Options
    # Putting creation options in the correct spot and preserving resolution
    warp_options = gdal.WarpOptions(
        format='GTiff',
        resampleAlg='near',
        xRes=orig_x_res,             # Enforces original pixel width
        yRes=orig_y_res,             # Enforces original pixel height
        creationOptions=[
            'COMPRESS=LZW',          # LZW compression
            'ZLEVEL=9',              # Maximum compression level
            'BIGTIFF=YES'            # Enable BigTIFF support
        ]
    )

    # Execute the Warp with the configured options object
    gdal.Warp(
        output_path, 
        tmp_ds, 
        options=warp_options
    )

    # Clean up memory buffers
    tmp_ds = None
    src_ds = None
    print(f"Raster successfully rotated and saved to: {output_path}")

# Example Usage:
# rotate_raster('input.tif', 'rotated_output.tif', 30.0, 500000.0, 4000000.0)
```

```python
from osgeo import ogr 
from shapely.wkt import loads as wkt_loads
from shapely.errors import TopologicalError
from rasterio import features
from rasterio.transform import Affine
```

```python
# Apply masks to EW, NS, PAR, PERP rasters
# Mask any correlation <= 0.5
# Mask anywhere vertical diff is > 1 m
areas = ['1','2','3','4','5N','5Sn','5Ss','6']
prefixes = ['one','two','three','four','fiveN','fiveS','fiveS','six']
areas = ['2A','2C','2DEF','5Sn','5Ss']
prefixes = ['two','two','two','fiveS','fiveS']

files = ['EWmicmac_actual_rotated.tif', 
         'NSmicmac_actual_rotated.tif' ,
         'ParallelDisp319.tif' ,
         'VerticalDisp_rotated.tif', 
         'PerpendicularDisp319.tif']

for arean,pref in zip(areas,prefixes):
    vertf = gdal.Open(f'{folder}MEC_Area{arean}/VerticalDisp_rotated.tif')
    vert = vertf.GetRasterBand(1).ReadAsArray()
    corr = gdal.Open(f'{folder}MEC_Area{arean}/Correlmicmac_rotated.tif')
    corr = corr.GetRasterBand(1).ReadAsArray()
    mask = ((np.absolute(vert) >= 1)|(corr <=0.5))

    # --- New: Erode shapefile polygons and create mask ---
    poly = f'/Volumes/ProjectsGarlockCSAF/CSAF_Lidar/GIS/FinalCutlines/{pref}.shp'
    driver = ogr.GetDriverByName('ESRI Shapefile')
    ds = driver.Open(poly, 0)
    layer = ds.GetLayer()

    # Get raster georeference info
    transform = transform = Affine.from_gdal(*vertf.GetGeoTransform())
    width = vertf.RasterXSize
    height = vertf.RasterYSize

    eroded_shapes = []
    for feature in layer:
        geom = feature.GetGeometryRef()
        if geom is None:
            continue
        try:
            shapely_geom = wkt_loads(geom.ExportToWkt())
            eroded = shapely_geom.buffer(-1) # buffer to avoid raster edge effects
            if eroded.is_valid and not eroded.is_empty:
                eroded_shapes.append((eroded, 1))
        except TopologicalError:
            continue

    # Rasterize eroded polygons
    polygon_mask = features.rasterize(
        eroded_shapes,
        out_shape=(height, width),
        transform=transform,
        fill=0,
        dtype=np.uint8
    ).astype(bool)

    # Combine: mask outside eroded area or failing data condition
    mask = mask | ~polygon_mask
    for file in files:
        imf = gdal.Open(f'{folder}MEC_Area{arean}/{file}')
        im = imf.GetRasterBand(1).ReadAsArray()
        im[mask] = -9999
        # remove median
        #im[im!=-9999] = im[im!=-9999] - np.median(im[im!=-9999])
        print(arean,file)
        tt.save_geotiff(im,f'{folder}MEC_Area{arean}/{file[:-4]}_masked.tif',imf.GetGeoTransform(),imf.GetProjection(),nodata=-9999)
```
